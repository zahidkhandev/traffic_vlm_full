# Run AutoQC Experiment Matrix (All 4 Inference Modes)

This notebook runs all supported inference modes in one go via `run_cli.py experiment`.


```python
from pathlib import Path
import json
import subprocess
import sys

ROOT = Path("Users/nkz3kor/qwen-autoqc")
CONFIG_PATH = ROOT / "configs/experiment.yaml"
SUMMARY_PATH = ROOT / "runs/experiments/experiment_summary.json"

print("Using config:", CONFIG_PATH)
print("Expected summary:", SUMMARY_PATH)
assert CONFIG_PATH.exists(), f"Missing config: {CONFIG_PATH}"
```

```python
# Run all experiment modes in a single command
cmd = [sys.executable, str(ROOT / "run_cli.py"), "experiment", "--config", str(CONFIG_PATH)]
print("Running:", " ".join(cmd))

proc = subprocess.run(cmd, cwd=str(ROOT), capture_output=True, text=True)
print("Return code:", proc.returncode)
print("\nSTDOUT:\n", proc.stdout)
if proc.stderr.strip():
    print("\nSTDERR:\n", proc.stderr)

if proc.returncode != 0:
    raise RuntimeError("Experiment matrix run failed. Check stdout/stderr above.")
```

```python
# Load summary generated across all modes
if not SUMMARY_PATH.exists():
    raise FileNotFoundError(f"Expected summary not found: {SUMMARY_PATH}")

summary = json.loads(SUMMARY_PATH.read_text(encoding="utf-8"))
print(json.dumps(summary, indent=2))
```
